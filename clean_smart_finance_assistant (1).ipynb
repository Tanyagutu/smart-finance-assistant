### Foundation Data Processing Functions

````
# 🤖 AI Collaboration: Data Loading and Cleaning Function
# Ask AI to help you create a robust data loading function

def load_and_clean_transaction_data(file_path):
    """
    Load and clean transaction data for the Smart Finance Assistant

    🤖 AI Collaboration Prompt:
    "Create a function to load CSV transaction data with Date, Amount, Category,
    Description columns. Handle dollar signs in Amount, missing values, and
    data validation. Include clear business-focused error messages."

    Args:
        file_path: Path to CSV file or file object
    Returns:
        pandas.DataFrame: Cleaned transaction data
    """
import pandas as pd

def load_and_clean_transaction_data(file_path):
    """
    Load and clean transaction data for the Smart Finance Assistant.
    """

    try:
        transactions_df = pd.read_csv(file_path)

        original_row_count = len(transactions_df)

        required_columns = ["Date", "Amount", "Category", "Description"]

        for column in required_columns:
            if column not in transactions_df.columns:
                return f"Error: Missing required column: {column}"

        # Count missing values before cleaning
        missing_values_before = transactions_df.isnull().sum().to_dict()

        # Fill missing text values
        transactions_df["Date"] = transactions_df["Date"].fillna("Unknown Date")
        transactions_df["Category"] = transactions_df["Category"].fillna("Uncategorised")
        transactions_df["Description"] = transactions_df["Description"].fillna("No Description")

        # Clean Amount column
        transactions_df["Amount"] = (
            transactions_df["Amount"]
            .astype(str)
            .str.replace("$", "", regex=False)
            .str.replace(",", "", regex=False)
            .str.strip()
        )

        # Convert Amount to numeric
        transactions_df["Amount"] = pd.to_numeric(
            transactions_df["Amount"],
            errors="coerce"
        )

        # Remove invalid amount rows
        transactions_df = transactions_df.dropna(subset=["Amount"])

        cleaned_row_count = len(transactions_df)
        removed_rows = original_row_count - cleaned_row_count

        if transactions_df.empty:
            return "Error: No valid transaction records found."

        transactions_df = transactions_df.reset_index(drop=True)

        cleaning_summary = {
            "Original Rows": original_row_count,
            "Cleaned Rows": cleaned_row_count,
            "Rows Removed": removed_rows,
            "Missing Values Before Cleaning": missing_values_before
        }

        print("✅ Transaction data loaded and cleaned successfully.")
        print("Cleaning Summary:", cleaning_summary)

        return transactions_df

    except FileNotFoundError:
        return "Error: The CSV file could not be found."

    except pd.errors.EmptyDataError:
        return "Error: The CSV file is empty."

    except Exception as error:
        return f"Error while loading transaction data: {error}"
# Test your function
# clean_data = load_and_clean_transaction_data(df_sample)
# print(clean_data)
````

````
# 🤖 AI Collaboration: Spending Analysis Function
# Ask AI to help you create comprehensive spending analysis

def analyze_spending_patterns(df):
    """
    Analyze spending patterns and generate business insights

    🤖 AI Collaboration Prompt:
    "Create a function that analyzes spending by category, calculates percentages,
    identifies top spending areas, and generates actionable financial insights
    formatted for business presentation."

    Args:
        df: Cleaned transaction DataFrame
    Returns:
        dict: Analysis results and insights
    """
def analyze_spending_patterns(df):
    """
    Analyze spending patterns and generate business insights.

    Args:
        df: Cleaned transaction DataFrame

    Returns:
        dict: Analysis results and insights
    """

    try:
        # Make a copy so the original data is not changed
        transactions_df = df.copy()

        # Fill missing categories so they are still included in the analysis
        transactions_df["Category"] = transactions_df["Category"].fillna("Uncategorised")

        # Separate spending and refunds
        spending_df = transactions_df[transactions_df["Amount"] > 0]
        refunds_df = transactions_df[transactions_df["Amount"] < 0]

        if spending_df.empty:
            return {"Error": "No spending transactions available for analysis."}

        # Calculate totals
        total_spending = spending_df["Amount"].sum()
        total_refunds = refunds_df["Amount"].sum()
        net_spending = transactions_df["Amount"].sum()

        # Group spending by category
        category_totals = spending_df.groupby("Category")["Amount"].sum().sort_values(ascending=False)

        # Calculate percentages
        category_percentages = ((category_totals / total_spending) * 100).round(2)

        # Identify top spending category
        top_category = category_totals.idxmax()
        top_category_total = category_totals.max()
        top_category_percentage = category_percentages[top_category]

        # Generate clearer business insight
        financial_insight = (
            f"The highest spending category was {top_category}, "
            f"with ${top_category_total:.2f}, representing "
            f"{top_category_percentage}% of total spending."
        )

        savings_recommendation = (
            f"Review the {top_category} category to see if any spending can be reduced."
        )

        # Return analysis results
        analysis_results = {
            "Total Spending": round(total_spending, 2),
            "Total Refunds": round(total_refunds, 2),
            "Net Spending": round(net_spending, 2),
            "Category Totals": category_totals.to_dict(),
            "Category Percentages": category_percentages.to_dict(),
            "Top Spending Category": top_category,
            "Top Category Spending": round(top_category_total, 2),
            "Top Category Percentage": round(top_category_percentage, 2),
            "Financial Insight": financial_insight,
            "Savings Recommendation": savings_recommendation
        }

        return analysis_results

    except Exception as error:
        return {"Error": f"Analysis failed: {error}"}

# Test your function
# analysis = analyze_spending_patterns(clean_data)
# print(analysis)
````

````
# 🤖 AI Collaboration: Business Insights Generator
# Ask AI to help create professional financial recommendations

def generate_financial_recommendations(analysis_data):
    """
    Generate actionable financial recommendations based on spending analysis

    🤖 AI Collaboration Prompt:
    "Based on spending analysis data, create professional financial
    recommendations. Include specific savings opportunities, spending
    pattern observations, and actionable advice formatted for a
    personal finance app user."

    Args:
        analysis_data: Dictionary with spending analysis results
    Returns:
        str: Formatted recommendations report
    """
def generate_financial_recommendations(analysis_data):
    """
    Generate actionable financial recommendations based on spending analysis.

    Args:
        analysis_data: Dictionary with spending analysis results

    Returns:
        str: Formatted recommendations report
    """

    try:
        if not analysis_data:
            return "Error: No analysis data was provided."

        total_spending = analysis_data.get("Total Spending", 0)
        total_refunds = analysis_data.get("Total Refunds", 0)
        net_spending = analysis_data.get("Net Spending", total_spending)
        top_category = analysis_data.get("Top Spending Category", "Unknown")
        top_category_percentage = analysis_data.get("Top Category Percentage", 0)

        category_advice = {
            "Food": "Try planning meals ahead or reducing takeaway purchases.",
            "Groceries": "Compare prices and make a shopping list before buying.",
            "Transport": "Look for cheaper travel options such as public transport or carpooling.",
            "Entertainment": "Set a monthly entertainment limit to avoid overspending.",
            "Shopping": "Avoid impulse purchases and focus on essential items first.",
            "Education": "This may be an important investment, but it should still be planned carefully.",
            "Subscription": "Review subscriptions and cancel any services you do not use often."
        }

        specific_advice = category_advice.get(
            top_category,
            "Review this category and check whether any expenses can be reduced."
        )

        report = "FINANCIAL RECOMMENDATIONS REPORT\n"
        report += "-" * 40 + "\n"
        report += f"Total Spending: ${total_spending:.2f}\n"
        report += f"Total Refunds: ${total_refunds:.2f}\n"
        report += f"Net Spending: ${net_spending:.2f}\n"
        report += f"Highest Spending Category: {top_category}\n\n"

        report += "Spending Pattern Observation:\n"
        report += (
            f"Your highest spending category was {top_category}, "
            f"which made up {top_category_percentage}% of your total spending.\n\n"
        )

        report += "Category-Specific Recommendation:\n"
        report += specific_advice + "\n"

        if total_refunds < 0:
            report += "\nRefund Note:\n"
            report += (
                f"Refunds reduced your overall spending by ${abs(total_refunds):.2f}. "
                "This shows why refunds should be tracked separately from normal expenses.\n"
            )

        report += "\nActionable Advice:\n"
        report += "- Review your highest spending category weekly\n"
        report += "- Set a realistic monthly spending limit\n"
        report += "- Separate essential and non-essential expenses\n"
        report += "- Track refunds separately so your net spending stays accurate\n"

        return report

    except Exception as error:
        return f"Error while generating recommendations: {error}"

# Test your function
# recommendations = generate_financial_recommendations(analysis)
# print(recommendations)
````

## RAG System for Financial Documents

````
# 📚 Simple RAG-style Finance Retrieval

def setup_financial_rag():
    """
    Set up simple finance documents for retrieval.
    """

    documents = [
        {
            "topic": "budget",
            "content": "A budget helps users divide income between needs, wants, and savings. Students can use budgeting to manage food, transport, education, and entertainment spending."
        },
        {
            "topic": "entertainment",
            "content": "Entertainment spending is usually flexible. Users can reduce entertainment costs by setting a monthly limit and tracking activities such as movies, games, and outings."
        },
        {
            "topic": "food",
            "content": "Food spending can increase quickly through takeaway, coffee, and campus meals. Meal planning and grocery shopping can help reduce food expenses."
        },
        {
            "topic": "saving",
            "content": "A savings goal helps users plan for future expenses. Saving a fixed amount weekly or monthly can make large goals easier to reach."
        },
        {
            "topic": "refund",
            "content": "Refunds reduce net spending and should be tracked separately from normal spending so the financial summary stays accurate."
        },
        {
            "topic": "transport",
            "content": "Transport costs can be managed by planning trips, using student discounts, public transport, or sharing rides where possible."
        }
    ]

    return documents


def ask_financial_rag(question, documents):
    """
    Retrieve a relevant finance document and return an answer.
    """

    if not question:
        return "Please enter a finance question."

    question_lower = question.lower()

    for document in documents:
        if document["topic"] in question_lower:
            return (
                f"Retrieved topic: {document['topic']}\n\n"
                f"{document['content']}"
            )

    return (
        "I could not find an exact matching finance document, but this finance assistant "
        "can currently retrieve information about budgeting, entertainment, food, saving, "
        "refunds, and transport."
    )
````

````
# 🤖 AI Collaboration: Custom Tool Development
# Ask AI to help you create useful financial calculators

from hands_on_ai import agent

def create_savings_calculator_tool():
    """
    Create a custom savings goal calculator as an agent tool

    🤖 AI Collaboration Prompt:
    "Create a savings goal calculator function that takes current savings,
    monthly contribution, and target amount, then calculates time to reach
    goal. Format output for user-friendly display."
    """
from hands_on_ai import agent
import math

def create_savings_calculator_tool():
    """
    Create a custom savings goal calculator as an agent tool.
    """

    def savings_calculator(current_savings, monthly_contribution, target_amount, extra_monthly_contribution=0):
        try:
            current_savings = float(current_savings)
            monthly_contribution = float(monthly_contribution)
            target_amount = float(target_amount)
            extra_monthly_contribution = float(extra_monthly_contribution)

            if current_savings < 0 or monthly_contribution < 0 or extra_monthly_contribution < 0 or target_amount <= 0:
                return "Error: Please enter valid positive numbers."

            if current_savings >= target_amount:
                return "You have already reached your savings goal."

            total_monthly_saving = monthly_contribution + extra_monthly_contribution

            if total_monthly_saving <= 0:
                return "Error: Monthly contribution must be greater than zero."

            remaining_amount = target_amount - current_savings
            months_needed = math.ceil(remaining_amount / total_monthly_saving)

            report = "SAVINGS GOAL CALCULATOR\n"
            report += "------------------------\n"
            report += f"Current Savings: ${current_savings:.2f}\n"
            report += f"Monthly Contribution: ${monthly_contribution:.2f}\n"
            report += f"Extra Monthly Contribution: ${extra_monthly_contribution:.2f}\n"
            report += f"Target Amount: ${target_amount:.2f}\n"
            report += f"Remaining Amount: ${remaining_amount:.2f}\n"
            report += f"Estimated Time to Goal: {months_needed} month(s)\n"

            if extra_monthly_contribution > 0 and monthly_contribution > 0:
                original_months = math.ceil(remaining_amount / monthly_contribution)
                months_saved = original_months - months_needed
                report += f"Time Saved With Extra Contributions: {months_saved} month(s)\n"
            else:
                report += "Tip: Increasing your monthly contribution can help you reach your goal faster.\n"

            return report

        except ValueError:
            return "Error: Please enter numeric values only."

    return savings_calculator


savings_calculator_tool = create_savings_calculator_tool()

print(savings_calculator_tool(200, 100, 1000, 50))

# Optional agent registration
# agent.register_tool("savings_calculator", savings_calculator_tool)

# Create the tool
savings_calculator_tool = create_savings_calculator_tool()

# Test the tool
print(savings_calculator_tool(200, 100, 1000, 50))

# Optional agent registration
# agent.register_tool("savings_calculator", savings_calculator_tool)
# Register your tool with the agent system
# agent.register_tool("savings_calculator", create_savings_calculator_tool)
````

````
import pandas as pd
import math
import os

# Hands-on-AI config
os.environ["HANDS_ON_AI_SERVER"] = "https://ollama.locollm.org"
os.environ["HANDS_ON_AI_MODEL"] = "gemma3:4b"
os.environ["HANDS_ON_AI_API_KEY"] = "Curtin2026ISYS20015002"

try:
    from hands_on_ai.chat import get_response
except:
    get_response = None


def load_and_clean_transaction_data(file_path):
    try:
        df = pd.read_csv(file_path)

        required_columns = ["Date", "Amount", "Category", "Description"]
        for column in required_columns:
            if column not in df.columns:
                return f"Error: Missing required column: {column}"

        df["Date"] = df["Date"].fillna("Unknown Date")
        df["Category"] = df["Category"].fillna("Uncategorised")
        df["Description"] = df["Description"].fillna("No Description")

        df["Amount"] = (
            df["Amount"]
            .astype(str)
            .str.replace("$", "", regex=False)
            .str.replace(",", "", regex=False)
            .str.strip()
        )

        df["Amount"] = pd.to_numeric(df["Amount"], errors="coerce")
        df = df.dropna(subset=["Amount"])

        if df.empty:
            return "Error: No valid transaction records found."

        return df.reset_index(drop=True)

    except Exception as error:
        return f"Error while loading transaction data: {error}"


def analyze_spending_patterns(df):
    try:
        df = df.copy()
        df["Category"] = df["Category"].fillna("Uncategorised")

        spending_df = df[df["Amount"] > 0]
        refunds_df = df[df["Amount"] < 0]

        if spending_df.empty:
            return {"Error": "No spending transactions available for analysis."}

        total_spending = spending_df["Amount"].sum()
        total_refunds = refunds_df["Amount"].sum()
        net_spending = df["Amount"].sum()

        category_totals = spending_df.groupby("Category")["Amount"].sum().sort_values(ascending=False)
        category_percentages = ((category_totals / total_spending) * 100).round(2)

        top_category = category_totals.idxmax()
        top_category_total = category_totals.max()
        top_category_percentage = category_percentages[top_category]

        return {
            "Total Spending": round(total_spending, 2),
            "Total Refunds": round(total_refunds, 2),
            "Net Spending": round(net_spending, 2),
            "Category Totals": category_totals.to_dict(),
            "Category Percentages": category_percentages.to_dict(),
            "Top Spending Category": top_category,
            "Top Category Spending": round(top_category_total, 2),
            "Top Category Percentage": round(top_category_percentage, 2)
        }

    except Exception as error:
        return {"Error": f"Analysis failed: {error}"}


def generate_financial_recommendations(analysis_data):
    if "Error" in analysis_data:
        return analysis_data["Error"]

    total_spending = analysis_data.get("Total Spending", 0)
    total_refunds = analysis_data.get("Total Refunds", 0)
    net_spending = analysis_data.get("Net Spending", total_spending)
    top_category = analysis_data.get("Top Spending Category", "Unknown")
    top_category_percentage = analysis_data.get("Top Category Percentage", 0)

    category_advice = {
        "Food": "Try planning meals ahead or reducing takeaway purchases.",
        "Groceries": "Compare prices and make a shopping list before buying.",
        "Transport": "Look for cheaper travel options such as public transport or carpooling.",
        "Entertainment": "Set a monthly entertainment limit to avoid overspending.",
        "Shopping": "Avoid impulse purchases and focus on essential items first.",
        "Education": "This may be important spending, but it should still be planned carefully.",
        "Subscription": "Review subscriptions and cancel services you do not use often."
    }

    specific_advice = category_advice.get(
        top_category,
        "Review this category and check whether any expenses can be reduced."
    )

    report = "FINANCIAL RECOMMENDATIONS REPORT\n"
    report += "-" * 40 + "\n"
    report += f"Total Spending: ${total_spending:.2f}\n"
    report += f"Total Refunds: ${total_refunds:.2f}\n"
    report += f"Net Spending: ${net_spending:.2f}\n"
    report += f"Highest Spending Category: {top_category}\n\n"
    report += f"Your highest spending category was {top_category}, which made up {top_category_percentage}% of total spending.\n\n"
    report += "Recommendation:\n"
    report += specific_advice + "\n"

    if total_refunds < 0:
        report += f"\nRefunds reduced your overall spending by ${abs(total_refunds):.2f}.\n"

    return report


def create_finance_chat_personality():
    return """
    You are a friendly Smart Finance Assistant for students and young adults.
    Give simple and practical advice about budgeting, saving, refunds,
    food spending, transport, subscriptions, and everyday student expenses.
    Be encouraging but realistic.
    """


def local_finance_response(message):
    message_lower = message.lower()

    if "budget" in message_lower:
        return "A simple student budget should separate needs like food, transport, and education from wants like shopping and entertainment."
    elif "saving" in message_lower or "save" in message_lower:
        return "Try setting a small weekly savings goal and checking your spending categories regularly."
    elif "refund" in message_lower:
        return "Refunds should be tracked separately because they reduce net spending."
    elif "coffee" in message_lower or "food" in message_lower:
        return "Food and coffee spending can add up quickly. Try setting a weekly limit or preparing some meals and drinks at home."
    elif "transport" in message_lower:
        return "Transport costs can be reduced by planning trips, using student discounts, or sharing rides where possible."
    elif "subscription" in message_lower:
        return "Review your subscriptions and cancel any services you do not use often."
    elif "entertainment" in message_lower:
        return "Entertainment is flexible spending, so setting a monthly limit can help avoid overspending."
    else:
        return "I can help with budgeting, saving, refunds, spending categories, subscriptions, and student finance habits."


def chatbot_response(message, history=None):
    if not message:
        return "Please enter a finance question."

    try:
        if get_response is not None:
            personality = create_finance_chat_personality()
            response = get_response(f"{personality}\n\nUser Question: {message}")
            if response:
                return response
    except:
        pass

    return local_finance_response(message)


def create_savings_calculator_tool():
    def savings_calculator(current_savings, monthly_contribution, target_amount, extra_monthly_contribution=0):
        try:
            current_savings = float(current_savings)
            monthly_contribution = float(monthly_contribution)
            target_amount = float(target_amount)
            extra_monthly_contribution = float(extra_monthly_contribution)

            if current_savings < 0 or monthly_contribution < 0 or extra_monthly_contribution < 0 or target_amount <= 0:
                return "Error: Please enter valid positive numbers."

            if current_savings >= target_amount:
                return "You have already reached your savings goal."

            total_monthly_saving = monthly_contribution + extra_monthly_contribution
            if total_monthly_saving <= 0:
                return "Error: Monthly contribution must be greater than zero."

            remaining_amount = target_amount - current_savings
            months_needed = math.ceil(remaining_amount / total_monthly_saving)

            return (
                "SAVINGS GOAL CALCULATOR\n"
                "------------------------\n"
                f"Current Savings: ${current_savings:.2f}\n"
                f"Monthly Contribution: ${monthly_contribution:.2f}\n"
                f"Extra Monthly Contribution: ${extra_monthly_contribution:.2f}\n"
                f"Target Amount: ${target_amount:.2f}\n"
                f"Remaining Amount: ${remaining_amount:.2f}\n"
                f"Estimated Time to Goal: {months_needed} month(s)\n"
            )

        except:
            return "Error: Please enter numeric values only."

    return savings_calculator


print("✅ Emergency functions loaded. Now run the Gradio UI cell.")
````
